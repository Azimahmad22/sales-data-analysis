```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create sample sales data
df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Headphones", "Laptop",
                "Phone", "Tablet", "Headphones", "Tablet",
                "Phone", "Laptop"],
    "Category": ["Electronics", "Electronics", "Accessories",
                 "Electronics", "Electronics", "Electronics",
                 "Accessories", "Electronics", "Electronics",
                 "Electronics"],
    "Region": ["North", "South", "East", "West", "North",
               "South", "West", "East", "South", "North"],
    "Quantity": [2, 5, 10, 1, 4, 3, 6, 2, 3, 2],
    "Price": [60000, 20000, 1500, 60000, 20000,
              25000, 1500, 25000, 20000, 60000]
})

# Data cleaning
df = df.drop_duplicates().copy()
df["Revenue"] = df["Quantity"] * df["Price"]

# Key performance indicators
print("Total Revenue:", df["Revenue"].sum())
print("Total Units Sold:", df["Quantity"].sum())
print("Average Order Revenue:", round(df["Revenue"].mean(), 2))

# Revenue by product
product_sales = df.groupby("Product")["Revenue"].sum()
print("\nRevenue by Product:")
print(product_sales.sort_values(ascending=False))

# Revenue by region
region_sales = df.groupby("Region")["Revenue"].sum()
print("\nRevenue by Region:")
print(region_sales.sort_values(ascending=False))

# Visualize revenue by product
product_sales.sort_values().plot(
    kind="barh", color="steelblue"
)
plt.title("Total Revenue by Product")
plt.xlabel("Revenue")
plt.tight_layout()
plt.show()

# Visualize revenue by region
sns.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Revenue by Region")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()

display(df)
```
